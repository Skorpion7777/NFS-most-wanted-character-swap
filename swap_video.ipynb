```python
from pathlib import Path

folder = Path("/content/drive/MyDrive/higgsfield")
content = folder / "content"
for path in (folder / "swap_video.py", content / "original.mp4",
             content / "basti.jpg", content / "fiat500.jpg"):
    if not path.is_file():
        raise FileNotFoundError(f"Missing: {path}")
    print(f"Found: {path.relative_to(folder)}")
```

```python
import os
from getpass import getpass

key_id = getpass("Higgsfield API key ID: ").strip()
key_secret = getpass("Higgsfield API key secret: ").strip()
if not key_id or not key_secret:
    raise ValueError("Both the API key ID and secret are required.")
for name in ("HF_KEY", "HF_API_KEY", "HF_API_SECRET"):
    os.environ.pop(name, None)
os.environ["HF_KEY"] = f"{key_id}:{key_secret}"
del key_id, key_secret
print("Credentials set for this session.")
```

```python
import subprocess
import sys

# Use a fresh output name for every intentional new generation.
output = folder / "outputs" / "basti_fiat500_480p_no_text.mp4"
command = [
    sys.executable, str(folder / "swap_video.py"),
    "--video", str(content / "original.mp4"),
    "--character", str(content / "basti.jpg"),
    "--car", str(content / "fiat500.jpg"),
    # No --duration: edit the full source clip (the API caps it at 30 seconds).
    "--resolution", "480p",
    "--output", str(output),
]
completed = subprocess.run(command, check=False)
if completed.returncode:
    raise RuntimeError(f"swap_video.py failed (exit {completed.returncode}); see Error above.")
print(f"Saved: {output}")
```
